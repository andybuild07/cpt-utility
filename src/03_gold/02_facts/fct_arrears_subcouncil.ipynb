# 1. Creating Table

## 1.1 Adding Constraints

```python
%sql
CREATE TABLE IF NOT EXISTS cpt_utility_catalog.gold.fct_arrears_subcouncil(
    arrears_key BIGINT NOT NULL,
    date_key BIGINT NOT NULL,
    subcouncil STRING NOT NULL,
    ageing_bucket_days INT NOT NULL,
    amount DECIMAL(13,2),

    CONSTRAINT pk_fct_arrears_subcouncil PRIMARY KEY(arrears_key) RELY,
    CONSTRAINT fk_fct_arrears_subcouncil_date FOREIGN KEY(date_key) REFERENCES cpt_utility_catalog.gold.dim_date(date_key) RELY
)
```

## 1.2 Populating Table

```python
%sql
INSERT OVERWRITE TABLE cpt_utility_catalog.gold.fct_arrears_subcouncil
SELECT
xxhash64(s.id) AS arrears_key,
COALESCE(CAST(date_format(s.date, 'yyyyMMdd') AS INT) , -1) AS date_key,
s.subcouncil AS subcouncil,
s.ageing_bucket_days AS ageing_bucket_days,
COALESCE(s.amount, 0.00) AS amount
FROM cpt_utility_catalog.silver.silver_subcouncils_arrears_cleaned s
```
